# Personal Debit Transaction Analysis

```python
import numpy as np
import pandas as pd
from pathlib import Path
from IPython.display import display
```

## Import and clean NAB transaction export

```python
# Load NAB transaction data from CSV files in the "data/nab" directory
nab_data_dir = Path("data/nab")
nab_csv_files = sorted(nab_data_dir.glob("*.csv"))

if not nab_csv_files:
    raise FileNotFoundError(f"No CSV files found in {nab_data_dir.resolve()}")

nab_transactions_by_file = {
    csv_file.stem: pd.read_csv(csv_file)
    for csv_file in nab_csv_files
}

# Check for missing or empty "Category" values in each DataFrame
print("Checking for missing or empty 'Category' values in NAB transaction files...\n")
for filename, df in nab_transactions_by_file.items():
    missing_category = (
        df["Category"].isna()
        | df["Category"].astype("string").str.strip().eq("").fillna(False)
    )

    if missing_category.any():
        print(filename)
        display(
            df.loc[missing_category]
        )
print("\nFinished checking for missing or empty 'Category' values.\n")
```

```python
# Format data
nab_transactions = pd.concat(
    nab_transactions_by_file.values(),
    ignore_index=True
)

# Pick only debit transactions
nab_transactions["Debit"] = nab_transactions["Amount"].apply(lambda x: -x if x < 0 else 0)

# nab_transactions.info()
nab_tranactions = nab_transactions.dropna(subset=['Date', 'Debit'])

display(nab_transactions.head())
nab_transactions.shape
```

## Import and clean BOM transaction export

```python
# Load BOM transaction data from CSV files in the "data/bom" directory
bom_data_dir = Path("data/bom")
bom_csv_files = sorted(bom_data_dir.glob("*.csv"))

if not bom_csv_files:
    raise FileNotFoundError(f"No CSV files found in {bom_data_dir.resolve()}")

bom_transactions_by_file = {
    csv_file.stem: pd.read_csv(csv_file)
    for csv_file in bom_csv_files
}

# display(bom_transactions_by_file[0].head())

#Check for missing or empty "SubCategory" values in each DataFrame
print("Checking for missing or empty 'SubCategory' values in NAB transaction files...\n")
for filename, df in bom_transactions_by_file.items():
    missing_category = (
        df["SubCategory"].isna()
        | df["SubCategory"].astype("string").str.strip().eq("").fillna(False)
    )

    if missing_category.any():
        print(filename)
        display(
            df.loc[missing_category]
        )
print("\nFinished checking for missing or empty 'SubCategory' values.\n")
```

```python
# Format data
bom_transactions = pd.concat(
    bom_transactions_by_file.values(),
    ignore_index=True
)

# Pick only debit transactions
bom_transactions = bom_transactions.dropna(subset=['Date', 'Debit'], how='any')

display(bom_transactions.head())
bom_transactions.shape

```
